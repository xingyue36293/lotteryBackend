````python
#https://www.168yyy.net

'''彩种 ID 对照表
ID	彩种名称	角标	玩法类别
101	重庆欢乐生肖	重庆	时时彩
103	广东快乐十分	广东	快乐十分
107	北京赛车(PK10)	北京	赛车
108	极速时时彩	极速	时时彩
109	澳洲幸运5	澳洲	时时彩
131	澳洲幸运8	澳洲	快乐十分
132	澳门幸运8	澳门	快乐十分
135	重庆幸运农场	重庆	快乐十分
170	极速飞艇	极速	赛车
171	168幸運飛艇	16	赛车
172	极速赛车	极速	赛车
175	澳洲幸运10	澳洲	赛车
200	极速六合彩	极速	六合彩
201	澳门六合彩5分	澳门	六合彩
202	新台湾六合彩	新台	六合彩
301	新澳门六合彩	新澳	六合彩'''

!curl -s "https://www.168yyy.net/api/lottery.php?action=history&gid=172&limit=50"
'''
# 极速赛车 历史开奖 API 接口文档

## 接口地址

```
GET https://www.168yyy.net/api/lottery.php
```

## 请求参数

| 参数名 | 类型 | 必填 | 说明 | 示例 |
|--------|------|------|------|------|
| `action` | String | 是 | 操作类型，查询历史列表用 `history` | `history` |
| `gid` | Integer | 是 | 游戏 ID（彩种编号） | `172`（极速赛车） |
| `limit` | Integer | 否 | 返回记录条数，默认约 50 | `50` |

> **彩种映射说明**：这个站的 `gid` 与之前 `api.apiote122.com` 的 `lotCode` 是**两套独立编号**，不能直接混用。`gid=172` 是本站的“极速赛车”，与 `lotCode=10037`（极速赛车）含义相同但编号不同。

## 返回结构

### 顶层

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `ok` | Boolean | 请求是否成功 | `true` |
| `server_time` | String | 服务器时间（ISO 8601 带时区） | `"2026-09-12T15:41:06+08:00"` |
| `game` | Object | 游戏元信息 | 见下 |
| `data` | Array | 开奖记录数组（时间倒序） | 见下 |

### `game` 对象 —— 游戏元信息

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `gid` | Integer | 游戏 ID | `172` |
| `name` | String | 游戏名称 | `"极速赛车"` |
| `type` | String | 类型标识 | `"race"` |
| `type_label` | String | 类型中文标签 | `"赛车"` |
| `fenlei` | String | 分类编号 | `"107"` |
| `ball_count` | Integer | 球数（注意：极速赛车实际是 10 个名次） | `10` |
| `sort` | Integer | 排序权重 | `4` |
| `latest` | Object | 最新一期信息 | 见下 |
| `next` | Object | 下一期预告 | 见下 |

### `game.latest` 对象 —— 最新一期

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `issue` | String | 期号 | `"34151873"` |
| `date` | String | 日期 | `"2026-09-12"` |
| `draw_time` | String | 开奖时间 | `"2026-09-12 15:40:33"` |
| `balls` | Array\<Int\> | 开奖号码（10 个名次） | `[4,9,7,10,3,2,8,1,5,6]` |
| `summary` | Object | 开奖摘要 | 见下 |

### `game.next` 对象 —— 下一期预告

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `issue` | String | 下期期号 | `"34151874"` |
| `draw_time` | String | 预计开奖时间 | `"2026-09-12 15:41:48"` |

### `summary` 对象 —— 开奖摘要

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `sum` | Integer | **冠亚军号码之和**（第 1 + 第 2 名） | `13` |
| `odd_even` | String | 冠亚和单双（"单" / "双"） | `"单"` |
| `big_small` | String | 冠亚和大小（"大" / "小" / "和"） | `"大"` |
| `dragon_tiger` | String | **冠军与第十名的龙虎**（"龙" / "虎" / "和"） | `"虎"` |
| `dragon_tigers` | Array\<String\> | **五个龙虎位置**的数组 | `["虎","龙","龙","龙","龙"]` |

### `data[]` 数组元素

每条记录的字段与 `game.latest` **完全一致**：

```json
{
  "issue": "34151873",
  "date": "2026-09-12",
  "draw_time": "2026-09-12 15:40:33",
  "balls": [4, 9, 7, 10, 3, 2, 8, 1, 5, 6],
  "summary": {
    "sum": 13,
    "odd_even": "单",
    "big_small": "大",
    "dragon_tiger": "虎",
    "dragon_tigers": ["虎", "龙", "龙", "龙", "龙"]
  }
}
```

## 关键规则说明（极速赛车）

### 1. 号码含义

`balls` 数组里是 **1~10 的名次**（不是普通号码），代表 10 辆赛车/赛马的名次排列。

```
balls = [4, 9, 7, 10, 3, 2, 8, 1, 5, 6]
         ↑冠军       ↑亚军  ↑第十名
```

### 2. `sum`（冠亚和）

= `balls[0]` + `balls[1]`，例如 `4 + 9 = 13`

- 范围：`3 ~ 19`
- `odd_even`：`sum` 为奇数返回"单"，偶数返回"双"
- `big_small`：
  - `sum ∈ [3, 11]` → `"小"`
  - `sum ∈ [12, 19]` → `"大"`
  - `sum == 11` 时视规则不同可能为 `"和"`（本数据里 11 出现过 `"和"`，说明采用 11 为和）

> 观察样本：`sum=11` 的记录 `big_small` 为 `"和"`，说明本站规则是 **`11` 为和**，`3~10` 为小，`12~19` 为大。

### 3. `dragon_tiger`（龙虎）

比较**冠军（第1名）** 与 **第十名** 的**名次数字大小**：

- 冠军名次数字 < 第十名名次数字 → `"龙"`
- 冠军名次数字 > 第十名名次数字 → `"虎"`
- 相等 → `"和"`

> 注意：这是**名次数字**比较，不是先后比较。数字小的排前面是"龙"。

### 4. `dragon_tigers`（五个龙虎位置）

数组 5 个元素，分别对应：

| 索引 | 比较对象 |
|------|---------|
| `[0]` | 第 1 名 vs 第 10 名 |
| `[1]` | 第 2 名 vs 第 9 名 |
| `[2]` | 第 3 名 vs 第 8 名 |
| `[3]` | 第 4 名 vs 第 7 名 |
| `[4]` | 第 5 名 vs 第 6 名 |

规则同上：前一位数字小为"龙"，大为"虎"。

## Python 调用示例

```python
import requests
import truststore
import json

truststore.inject_into_ssl()

URL = "https://www.168yyy.net/api/lottery.php"

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:155.0) Gecko/20100101 Firefox/155.0",
    "Accept": "*/*",
    "Accept-Language": "zh-CN,zh;q=0.9,en;q=0.8",
    "Referer": "https://www.168yyy.net/",
}


def fetch_history(gid=172, limit=50):
    params = {"action": "history", "gid": gid, "limit": limit}
    resp = requests.get(URL, params=params, headers=HEADERS, timeout=10)
    resp.raise_for_status()
    return resp.json()


if __name__ == "__main__":
    data = fetch_history(gid=172, limit=50)

    if not data.get("ok"):
        print("请求失败")
        raise SystemExit

    game = data["game"]
    print(f"游戏: {game['name']} (gid={game['gid']})")
    print(f"球数: {game['ball_count']}")
    print(f"服务器时间: {data['server_time']}")
    print(f"\n最新一期: {game['latest']['issue']} @ {game['latest']['draw_time']}")
    print(f"  号码: {game['latest']['balls']}")
    print(f"  冠亚和: {game['latest']['summary']['sum']} "
          f"({game['latest']['summary']['odd_even']} / {game['latest']['summary']['big_small']})")
    print(f"  龙虎: {game['latest']['summary']['dragon_tiger']}")
    print(f"\n下一期: {game['next']['issue']} @ {game['next']['draw_time']}")

    print(f"\n共获取 {len(data['data'])} 条历史记录，前 5 条：")
    for r in data["data"][:5]:
        s = r["summary"]
        print(f"  {r['issue']}  {r['draw_time']}  {r['balls']}  "
              f"和={s['sum']} {s['odd_even']}{s['big_small']}  龙虎={s['dragon_tiger']}")


### 实用分析片段

```python
from collections import Counter

records = data["data"]

# 1. 冠亚和单双分布
oe = Counter(r["summary"]["odd_even"] for r in records)
print(f"单: {oe['单']}  双: {oe['双']}")

# 2. 冠亚和大小分布
bs = Counter(r["summary"]["big_small"] for r in records)
print(f"大: {bs['大']}  小: {bs['小']}  和: {bs['和']}")

# 3. 冠军龙虎分布
dt = Counter(r["summary"]["dragon_tiger"] for r in records)
print(f"龙: {dt['龙']}  虎: {dt['虎']}  和: {dt['和']}")

# 4. 五个龙虎位置统计
for i in range(5):
    c = Counter(r["summary"]["dragon_tigers"][i] for r in records)
    print(f"位置{i+1}龙虎: 龙={c['龙']} 虎={c['虎']} 和={c['和']}")

# 5. 提取冠军号（第1名）
champions = [r["balls"][0] for r in records]
print("冠军号序列:", champions)

# 6. 判断某个冠军号是否连续出现
from itertools import groupby
runs = [(k, len(list(g))) for k, g in groupby(champions)]
print("冠军号连续段:", runs)
```

## 其他可能支持的 action 值

根据该接口的命名习惯（`action=history`），通常还会支持以下操作（需自行验证）：

| `action` 值 | 可能用途 |
|------------|---------|
| `history` | 历史开奖列表（已验证） |
| `latest` | 仅获取最新一期 |
| `next` | 获取下一期预告 |
| `games` | 获取所有游戏列表 |
| `trend` | 获取走势数据 |

## 与 `api.apiote122.com` 接口对照

| 维度 | `168yyy.net/api/lottery.php` | `api.apiote122.com/CQShiCai/*` |
|------|------------------------------|--------------------------------|
| 彩种标识 | `gid=172` | `lotCode=10037` |
| 号码格式 | 数组 `[4,9,7,10,3,2,8,1,5,6]` | 字符串 `"4,9,7,10,3,2,8,1,5,6"` |
| 单双/大小 | 中文 `"单"/"双"` | 数字 `1/0` |
| 龙虎 | 中文 `"龙"/"虎"/"和"` | 数字 `0/1/2` |
| 冠亚和范围 | `3~19`，`11` 为和 | 视彩种而定 |
| 游戏元信息 | 有 `game` 字段 | 需另外接口查 |
| 下一期预告 | 直接内置在 `game.next` | 在 `getBaseCQShiCai` 里 |
| 分页 | 支持 `limit` | 无 |

## 注意事项

1. **期号是字符串**：`issue` 在 JSON 里是字符串（带引号），不是数字，解析时别当 int 处理。
2. **`limit` 上限**：实测 `limit=50` 可以，更大的值（如 200、500）需自行测试。
3. **`big_small` 可能出现 `"和"`**：冠亚和 `= 11` 时为"和"，做统计时要留出这个分支。
4. **`balls` 里是名次数字（1~10）**：不是普通号码，两个"名次"比较大小才是龙虎逻辑。
5. **跨域**：这个 API 域名是 `www.168yyy.net`，你之前用 `api.apiote122.com`，两者是**不同的数据源**，不要混淆。'''
````

````python
'''
彩票名称	代码	lotCode ID
福彩3D	fc3d	1
体彩排列3	tcpl3	2
体彩排列5	tcpl5	3
香港六合彩	sglh	4
澳门六合彩	amlh	5
新澳门六合彩	samlh	6
台湾五分彩	taiwanbg	7
币安比特分分彩	babtffc	8
币安以太分分彩	baytffc	9
奇趣腾讯分分彩	cctsffc	11
奇趣腾讯五分彩	cctswfc	12
奇趣腾讯十分彩	cctssfc	13
老河內五分彩	lhnwfc	14
澳洲幸运5	aozxy5	15
168极速赛车	jssc168	16
澳洲幸运10	aozxy10	17
168幸运飞艇	syft168	18
168极速飞艇	jsft168	19
广东11选5	gd11x5	20
曼谷11选5	bk11x5	21
上证五分彩	sjwfc	22
深证五分彩	szwfc	23
韩国28五分彩	hg28wfc	24
重庆时时彩	cqssc	25
哈希分分彩	hxffc	26
哈希五分彩	hxwfc	27
哈希十分彩	hxshfc	28
哈希三分彩	hxsfc	29
波场分分彩	bcffc	30
波场三分彩	bcsfc	31
波场五分彩	bcwfc	32
分分动物运动会	dwydhssc	33
三分动物运动会	dwydhsfc	34
奖杯动物运动会	jbdwydh	35
奖牌动物运动会	jpdwydh	36
'''

#!curl -s --compressed "https://api.chuanqiking.com/CQShiCai/getBaseCQShiCai?issue=202609121362&lotCode=26"
'''
# 哈希分分彩 开奖数据 API 接口文档

## 接口地址

```
GET https://api.chuanqiking.com/CQShiCai/getBaseCQShiCai
```

> 这是 `api.chuanqiking.com` 上的接口，与 `1688507.com`、`1688455.com` 属**不同数据源**。请求需支持 gzip 解压（建议加 `--compressed` 或用 `requests`）。

## 请求参数

| 参数名 | 类型 | 必填 | 说明 | 示例 |
|--------|------|------|------|------|
| `issue` | String | 是 | 期号（**字符串**，勿转 int） | `"202609121362"` |
| `lotCode` | Integer | 是 | 彩种编号 | `26`（哈希分分彩） |

> **彩种对照**：`lotCode=26` 对应 `lotCodeType=cqssc`。注意此站 `lotCode` 是 **1xxxx 系列**之外的编号体系，与 `10012`（澳洲幸运10）属于不同数据源。

## 返回结构

### 顶层

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `errorCode` | Integer | 错误码，0 表示成功 | `0` |
| `message` | String | 操作结果描述 | `"操作成功"` |
| `result` | Object | 业务数据包装 | 见下 |

### `result` 对象

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `businessCode` | Integer | 业务状态码，0 表示成功 | `0` |
| `message` | String | 业务描述 | `"操作成功"` |
| `data` | Object | 开奖数据详情 | 见下 |

### `data` 对象 —— 彩种与状态信息

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `serverTime` | String | 服务器当前时间 | `"2026-09-12 22:45:47"` |
| `lotCode` | **String** | 彩种编号（**注意是字符串**） | `"26"` |
| `lotName` | String | 彩种名称 | `"哈希分分彩"` |
| `iconUrl` | String | 彩种图标 | `""` |
| `totalCount` | Integer | 当天总期数 | `1440`（= 24×60，每分钟一期） |
| `shelves` | Integer | 是否上架（0=正常） | `0` |
| `groupCode` | Integer | 分组编码 | `2` |
| `frequency` | String | 频率（通常为空） | `""` |
| `lotteryStatus` | Integer | 彩种状态（0=正常开奖） | `0` |
| `category` | String | 类别 | `""` |
| `hot` | Integer | 热度标记 | `0` |
| `index` | Integer | 排序索引 | `100` |
| `id` | Object | **MongoDB ObjectId**（嵌套 `$oid`） | `{"$oid":"6aa56587100187c71d09aaf4"}` |
| `sdrawCount` | String | 空字符串 | `""` |
| `status` | Integer | 当前状态（0=等待开奖） | `0` |

> **重要**：`id` 是 MongoDB 的 ObjectId 包装格式，不是普通字符串。Python 中取用时需 `d["id"]["$oid"]`。

### `data` 对象 —— 上期开奖结果

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `preDrawIssue` | **String** | 上期期号（**注意是字符串**） | `"202609121365"` |
| `preDrawTime` | String | 上期开奖时间 | `"2026-09-12 22:45:12"` |
| `preDrawCode` | String | 上期开奖号码（5 个，逗号分隔） | `"9,5,7,0,8"` |
| `preDrawDate` | String | 上期开奖日期 | `"2026-09-12 00:00:00"` |
| `drawCount` | Integer | 当天已开期数 | `1365` |

### `data` 对象 —— 当前期/下期信息

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `drawIssue` | **String** | 当前/下期期号（字符串） | `"202609121366"` |
| `drawTime` | String | 预计开奖时间 | `"2026-09-12 22:46:00"` |

### `data` 对象 —— 各球位号码（1~5）

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `firstNum` | Integer | 第一球号码 | `9` |
| `secondNum` | Integer | 第二球号码 | `5` |
| `thirdNum` | Integer | 第三球号码 | `7` |
| `fourthNum` | Integer | 第四球号码 | `0` |
| `fifthNum` | Integer | 第五球号码 | `8` |

### `data` 对象 —— 总和与形态分析

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `sumNum` | Integer | 5 球号码之和 | `29` |
| `sumSingleDouble` | Integer | 总和单双（1=单，0=双） | `0` |
| `sumBigSmall` | Integer | 总和大小（1=大，0=小） | `0` |
| `dragonTiger` | Integer | 龙虎（0=虎，1=龙，2=和） | `0` |

### `data` 对象 —— 各球位单双大小

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `firstSingleDouble` | Integer | 第一球单双（1=单，0=双） | `0` |
| `firstBigSmall` | Integer | 第一球大小（1=大，0=小） | `0` |
| `secondSingleDouble` | Integer | 第二球单双 | `0` |
| `secondBigSmall` | Integer | 第二球大小 | `0` |
| `thirdSingleDouble` | Integer | 第三球单双 | `0` |
| `thirdBigSmall` | Integer | 第三球大小 | `0` |
| `fourthSingleDouble` | Integer | 第四球单双 | `1` |
| `fourthBigSmall` | Integer | 第四球大小 | `1` |
| `fifthSingleDouble` | Integer | 第五球单双 | `1` |
| `fifthBigSmall` | Integer | 第五球大小 | `0` |

> **大小规则**：`0~4` 为小，`5~9` 为大。
> **单双规则**：`1,3,5,7,9` 为单，`0,2,4,6,8` 为双。

### `data` 对象 —— 组合形态

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `lastThree` | Integer | 前三形态（见下表） | `1` |
| `betweenThree` | Integer | 中三形态 | `0` |
| `behindThree` | Integer | 后三形态 | `0` |

**形态枚举**：

| 值 | 含义 |
|----|------|
| `0` | 杂六 |
| `1` | 半顺 |
| `2` | 顺子 |
| `3` | 对子 |
| `4` | 豹子 |

## 完整返回示例

```json
{
  "errorCode": 0,
  "message": "操作成功",
  "result": {
    "businessCode": 0,
    "message": "操作成功",
    "data": {
      "serverTime": "2026-09-12 22:45:47",
      "lotCode": "26",
      "lotName": "哈希分分彩",
      "iconUrl": "",
      "totalCount": 1440,
      "shelves": 0,
      "groupCode": 2,
      "frequency": "",
      "lotteryStatus": 0,
      "category": "",
      "hot": 0,
      "index": 100,
      "id": {"$oid": "6aa56587100187c71d09aaf4"},
      "drawCount": 1365,
      "preDrawIssue": "202609121365",
      "preDrawTime": "2026-09-12 22:45:12",
      "preDrawCode": "9,5,7,0,8",
      "drawTime": "2026-09-12 22:46:00",
      "drawIssue": "202609121366",
      "sumNum": 29,
      "sumSingleDouble": 0,
      "sumBigSmall": 0,
      "dragonTiger": 0,
      "firstBigSmall": 0,
      "firstSingleDouble": 0,
      "secondBigSmall": 0,
      "secondSingleDouble": 0,
      "thirdBigSmall": 0,
      "thirdSingleDouble": 0,
      "fourthBigSmall": 1,
      "fourthSingleDouble": 1,
      "fifthBigSmall": 0,
      "fifthSingleDouble": 1,
      "behindThree": 0,
      "betweenThree": 0,
      "lastThree": 1,
      "preDrawDate": "2026-09-12 00:00:00",
      "firstNum": 9,
      "secondNum": 5,
      "thirdNum": 7,
      "fourthNum": 0,
      "fifthNum": 8,
      "sdrawCount": "",
      "status": 0
    }
  }
}
```

## Python 调用示例

```python
import requests
import truststore

truststore.inject_into_ssl()

URL = "https://api.chuanqiking.com/CQShiCai/getBaseCQShiCai"
LOT_CODE = 26  # 哈希分分彩

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:155.0) Gecko/20100101 Firefox/155.0",
    "Accept": "*/*",
    "Accept-Language": "zh-CN,zh;q=0.9,en;q=0.8",
    "Origin": "https://www.chuanqiking.com",
    "Referer": "https://www.chuanqiking.com/",
}


def fetch_ssc(issue, lot_code=26):
    params = {"issue": issue, "lotCode": lot_code}
    resp = requests.get(URL, params=params, headers=HEADERS, timeout=10)
    resp.raise_for_status()
    return resp.json()


if __name__ == "__main__":
    data = fetch_ssc(issue="202609121362", lot_code=LOT_CODE)
    if data["errorCode"] != 0:
        print(f"请求失败: {data['message']}")
        raise SystemExit

    d = data["result"]["data"]
    print(f"彩种: {d['lotName']} (lotCode={d['lotCode']})")
    print(f"当天总期数: {d['totalCount']}  已开期数: {d['drawCount']}")
    print(f"服务器时间: {d['serverTime']}\n")

    print(f"上期期号: {d['preDrawIssue']}")
    print(f"上期时间: {d['preDrawTime']}")
    print(f"上期号码: {d['preDrawCode']}")
    print(f"号码拆分: {d['firstNum']} {d['secondNum']} {d['thirdNum']} "
          f"{d['fourthNum']} {d['fifthNum']}")
    print(f"总和: {d['sumNum']}  "
          f"({'单' if d['sumSingleDouble']==1 else '双'} / "
          f"{'大' if d['sumBigSmall']==1 else '小'})")

    print(f"\n下期期号: {d['drawIssue']}  时间: {d['drawTime']}")

    # 各球位单双大小
    print("\n--- 各球位分析 ---")
    for i, (sd_key, bs_key) in enumerate([
        ("firstSingleDouble", "firstBigSmall"),
        ("secondSingleDouble", "secondBigSmall"),
        ("thirdSingleDouble", "thirdBigSmall"),
        ("fourthSingleDouble", "fourthBigSmall"),
        ("fifthSingleDouble", "fifthBigSmall"),
    ], 1):
        sd = "单" if d[sd_key] == 1 else "双"
        bs = "大" if d[bs_key] == 1 else "小"
        print(f"第{i}球: {sd} / {bs}")

    # 形态
    forms = {0: "杂六", 1: "半顺", 2: "顺子", 3: "对子", 4: "豹子"}
    print(f"\n前三形态: {forms.get(d['lastThree'], '未知')}")
    print(f"中三形态: {forms.get(d['betweenThree'], '未知')}")
    print(f"后三形态: {forms.get(d['behindThree'], '未知')}")
```

## 关键点与注意事项

1. **期号是字符串，不是整数**
   `preDrawIssue = "202609121365"` 和 `drawIssue = "202609121366"` 都是**字符串**，与 `1688507.com` 的整数期号不同。直接传参不要转 int，否则会丢失前导零并可能被后端拒绝。

2. **`lotCode` 也是字符串**
   `data.lotCode = "26"` 是字符串，但请求参数 `lotCode=26` 是整数——请求时按整数传，响应里是字符串，注意类型转换。

3. **`id` 是 MongoDB ObjectId**
   `id` 字段是 `{"$oid": "..."}` 形式，取值需 `d["id"]["$oid"]`，不能直接当字符串用。

4. **`preDrawDate` 是 `"YYYY-MM-DD 00:00:00"`**
   包含时间的日期格式，与 `preDrawTime` 有区别，做日期匹配时注意截取前 10 位。

5. **哈希分分彩 = 每分钟一期**
   `totalCount = 1440`，即一天 1440 期（24 × 60）。这是“分分彩”的典型特征。

6. **`drawCount` 是当天已开期数**
   示例中为 1365，对应 22:45 已过 1365 分钟（约 22:45 的开奖记录条数），可用于验证进度。

7. **与 `1688507.com` 的澳洲幸运10 对照**

| 维度 | `api.chuanqiking.com`（本接口） | `1688507.com` |
|------|-------------------------------|----------------|
| 彩种 | 哈希分分彩（lotCode=26） | 澳洲幸运10（lotCode=10012） |
| 期号类型 | **字符串** `"202609121365"` | **整数** `21358963` |
| 球数 | 5 球（0~9） | 10 个名次（1~10） |
| `lotCode` 类型 | 响应里是字符串 | 整数 |
| 数据 ID | MongoDB ObjectId | 无 |
| 日总期数 | 1440（每分钟） | 288（每 5 分钟） |

8. **必须支持压缩**：如遇 `422` 错误，加 `--compressed`（curl）或使用 `requests`（自动处理）。'''


#!curl -s --compressed "https://api.chuanqiking.com/CQShiCai/getBaseCQShiCaiList?lotCode=26"
'''
# 哈希分分彩 历史开奖列表 API 接口文档

## 接口地址

```
GET https://api.chuanqiking.com/CQShiCai/getBaseCQShiCaiList
```

> 与 `getBaseCQShiCai`（单期详情）同一数据源，但**返回多期列表**。请求需支持 gzip 解压（curl 加 `--compressed`，requests 自动处理），否则会返回 `422`。

## 请求参数

| 参数名 | 类型 | 必填 | 说明 | 示例 |
|--------|------|------|------|------|
| `lotCode` | Integer | 是 | 彩种编号 | `26`（哈希分分彩） |

> 未发现分页、日期、条数等参数。一次返回**当天全部已开期数**（示例中约 200 期，覆盖 20:02 ~ 23:21），按**时间倒序**排列，`data[0]` 是最新一期。

## 返回结构

### 顶层

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `errorCode` | Integer | 错误码，0 表示成功 | `0` |
| `message` | String | 操作结果描述 | `"操作成功"` |
| `result` | Object | 业务数据包装 | 见下 |

### `result` 对象

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `businessCode` | Integer | 业务状态码，0 表示成功 | `0` |
| `message` | String | 业务描述 | `"操作成功"` |
| `data` | Array | **开奖记录数组**（时间倒序） | 见下 |

### `data[]` 数组元素（单条记录）

| 字段 | 类型 | 说明 | 示例值 |
|------|------|------|--------|
| `preDrawIssue` | **String** | 期号（字符串） | `"202609121401"` |
| `preDrawTime` | String | 开奖时间 | `"2026-09-12 23:21:10"` |
| `preDrawCode` | String | 开奖号码（5 个，逗号分隔，**无前导零**） | `"3,5,1,5,1"` |
| `sumNum` | Integer | 5 球号码之和 | `15` |
| `sumSingleDouble` | Integer | 总和单双（1=单，0=双） | `0` |
| `sumBigSmall` | Integer | 总和大小（1=大，0=小） | `1` |
| `dragonTiger` | Integer | 龙虎（0=虎，1=龙，2=和） | `0` |
| `firstBigSmall` / `firstSingleDouble` | Integer | 第一球大小 / 单双 | `1` / `0` |
| `secondBigSmall` / `secondSingleDouble` | Integer | 第二球大小 / 单双 | `0` / `0` |
| `thirdBigSmall` / `thirdSingleDouble` | Integer | 第三球大小 / 单双 | `1` / `0` |
| `fourthBigSmall` / `fourthSingleDouble` | Integer | 第四球大小 / 单双 | `0` / `0` |
| `fifthBigSmall` / `fifthSingleDouble` | Integer | 第五球大小 / 单双 | `1` / `0` |
| `behindThree` | Integer | 后三形态（见下表） | `0` |
| `betweenThree` | Integer | 中三形态 | `3` |
| `lastThree` | Integer | 前三形态 | `3` |
| `groupCode` | Integer | 分组编码 | `2` |

> **注意**：与单期接口 `getBaseCQShiCai` 相比，列表接口**不返回** `firstNum ~ fifthNum`、`lotName`、`drawIssue`、`drawTime`、`id` 等字段，也没有 `serverTime`。需要这些信息，请调单期接口。

## 枚举值速查

### 单双 / 大小 / 龙虎

| 字段 | 取值 | 含义 |
|------|------|------|
| `sumSingleDouble`、`*SingleDouble` | `1` / `0` | 单 / 双 |
| `sumBigSmall`、`*BigSmall` | `1` / `0` | 大 / 小 |
| `dragonTiger` | `0` / `1` / `2` | 虎 / 龙 / 和 |

> **大小规则（单球）**：`0~4` 为小，`5~9` 为大。
> **总和大小**：各球相加后 `0~22` 为小，`23~45` 为大。
> **龙虎规则**：第一球 > 第五球为"龙"（`1`）；< 为"虎"（`0`）；= 为"和"（`2`）。

### 形态（`behindThree` / `betweenThree` / `lastThree`）

| 值 | 含义 |
|----|------|
| `0` | 杂六 |
| `1` | 半顺 |
| `2` | 顺子 |
| `3` | 对子 |
| `4` | 豹子 |

## 返回示例（截取前 2 条）

```json
{
  "errorCode": 0,
  "message": "操作成功",
  "result": {
    "businessCode": 0,
    "message": "操作成功",
    "data": [
      {
        "preDrawIssue": "202609121401",
        "preDrawTime": "2026-09-12 23:21:10",
        "preDrawCode": "3,5,1,5,1",
        "sumNum": 15,
        "sumSingleDouble": 0,
        "sumBigSmall": 1,
        "dragonTiger": 0,
        "firstBigSmall": 1,
        "firstSingleDouble": 0,
        "secondBigSmall": 0,
        "secondSingleDouble": 0,
        "thirdBigSmall": 1,
        "thirdSingleDouble": 0,
        "fourthBigSmall": 0,
        "fourthSingleDouble": 0,
        "fifthBigSmall": 1,
        "fifthSingleDouble": 0,
        "behindThree": 0,
        "betweenThree": 3,
        "lastThree": 3,
        "groupCode": 2
      },
      {
        "preDrawIssue": "202609121400",
        "preDrawTime": "2026-09-12 23:20:14",
        "preDrawCode": "8,1,2,6,5",
        "sumNum": 22,
        "sumSingleDouble": 1,
        "sumBigSmall": 1,
        "dragonTiger": 0,
        "firstBigSmall": 0,
        "firstSingleDouble": 1,
        "secondBigSmall": 1,
        "secondSingleDouble": 0,
        "thirdBigSmall": 1,
        "thirdSingleDouble": 1,
        "fourthBigSmall": 0,
        "fourthSingleDouble": 1,
        "fifthBigSmall": 0,
        "fifthSingleDouble": 0,
        "behindThree": 1,
        "betweenThree": 1,
        "lastThree": 1,
        "groupCode": 2
      }
    ]
  }
}
```

## Python 调用示例

```python
import requests
import truststore

truststore.inject_into_ssl()

URL = "https://api.chuanqiking.com/CQShiCai/getBaseCQShiCaiList"
LOT_CODE = 26

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:155.0) Gecko/20100101 Firefox/155.0",
    "Accept": "*/*",
    "Accept-Language": "zh-CN,zh;q=0.9,en;q=0.8",
    "Origin": "https://www.chuanqiking.com",
    "Referer": "https://www.chuanqiking.com/",
}


def fetch_history(lot_code=26):
    params = {"lotCode": lot_code}
    resp = requests.get(URL, params=params, headers=HEADERS, timeout=10)
    resp.raise_for_status()
    return resp.json()


if __name__ == "__main__":
    data = fetch_history(LOT_CODE)
    if data["errorCode"] != 0:
        print(f"请求失败: {data['message']}")
        raise SystemExit

    records = data["result"]["data"]
    print(f"共获取 {len(records)} 条历史记录\n")
    print(f"最新一期: {records[0]['preDrawIssue']} @ {records[0]['preDrawTime']}")
    print(f"最旧一期: {records[-1]['preDrawIssue']} @ {records[-1]['preDrawTime']}\n")

    print("--- 最近 5 期 ---")
    for r in records[:5]:
        sd = "单" if r["sumSingleDouble"] == 1 else "双"
        bs = "大" if r["sumBigSmall"] == 1 else "小"
        dt = {0: "虎", 1: "龙", 2: "和"}[r["dragonTiger"]]
        print(f"{r['preDrawIssue']}  {r['preDrawTime']}  {r['preDrawCode']}  "
              f"和={r['sumNum']}({sd}/{bs})  龙虎={dt}")
```

## 实用分析片段

```python
from collections import Counter

records = data["result"]["data"]

# 1. 冠亚和/总和单双分布
oe = Counter(r["sumSingleDouble"] for r in records)
print(f"单: {oe[1]}  双: {oe[0]}")

# 2. 总和大小分布
bs = Counter(r["sumBigSmall"] for r in records)
print(f"大: {bs[1]}  小: {bs[0]}")

# 3. 龙虎分布
dt = Counter(r["dragonTiger"] for r in records)
print(f"龙: {dt[1]}  虎: {dt[0]}  和: {dt[2]}")

# 4. 后三形态分布
forms = {0: "杂六", 1: "半顺", 2: "顺子", 3: "对子", 4: "豹子"}
bt = Counter(r["behindThree"] for r in records)
for k, v in bt.most_common():
    print(f"后三 {forms[k]}: {v}")

# 5. 提取冠军号码（第一球）
firsts = [int(r["preDrawCode"].split(",")[0]) for r in records]
print("第一球分布:", Counter(firsts))

# 6. 查找“豹子”（后三位相同）
for r in records:
    nums = r["preDrawCode"].split(",")
    if len(set(nums[-3:])) == 1:
        print(f"豹子: {r['preDrawIssue']} {r['preDrawCode']}")

# 7. 连续相同总和单双
from itertools import groupby
for k, g in groupby(records, key=lambda x: x["sumSingleDouble"]):
    cnt = len(list(g))
    if cnt >= 5:
        label = "单" if k == 1 else "双"
        print(f"连续 {cnt} 期 {label}")
```

## 与单期接口的对照

| 维度 | `getBaseCQShiCai`（单期） | `getBaseCQShiCaiList`（本接口） |
|------|--------------------------|-------------------------------|
| 用途 | 查指定期或最新一期详情 | 拉当天历史列表 |
| 请求参数 | `issue` + `lotCode` | 仅 `lotCode` |
| `data` 形态 | 对象 | 数组（时间倒序） |
| 球位字段 | 有 `firstNum~fifthNum` | **无**，只能从 `preDrawCode` split |
| 彩种信息 | 有 `lotName`、`totalCount` 等 | **无** |
| 下期预告 | 有 `drawIssue`、`drawTime` | **无** |
| `serverTime` | 有 | **无** |
| 数据量 | 单条 | 当天全部（示例约 200 条） |

## 注意事项

1. **期号是字符串**：`"202609121401"`，不要转 int，避免丢失语义。
2. **`preDrawCode` 无前导零**：与澳洲幸运10（`"10,08,03,..."`）不同，本接口号码是 `"3,5,1,5,1"` 这种普通数字，split 后直接 `int()` 即可。
3. **没有球位数值字段**：需要 `firstNum` 等，只能从 `preDrawCode` 解析。
4. **没有 `serverTime`**：判断数据新鲜度只能看 `records[0]["preDrawTime"]`。
5. **数据量随时间变化**：早上去请求可能只有几十条，晚上会累积到 1000+ 条（`totalCount=1440`），注意内存。
6. **必须支持 gzip 压缩**：否则 `422`。
7. **`dragonTiger` 有"和"值（2）**：做二分类统计时要注意排除或单独计数。
8. **列表里没有任何"期号范围"过滤参数**：如需特定期号，请用单期接口 `getBaseCQShiCai`。'''

````
